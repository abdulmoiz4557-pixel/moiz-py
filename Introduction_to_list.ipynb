**Accessing list elements**

```python
cities = ['singapore','newyork','islamabad']
numbers = [45,56,50,64,35]
print(cities[0])
print(numbers[1:4])
```

we can access the elements of lists using indexes/positions.
from above example a list named as cities  and other is numbers.
if we print the list cities but we want only specific elements from the list we will use index like if we write print(cities[0]) the element on index 0 will be printed like singapore was printed above . and the nuumbers are printed from index 1 to 4 excluding 4 because the ending point is always excluded while printing the elements of a list.

**Updating lists**

```python
cities = ['singapore','newyork','islamabad']
print(cities)
print("orignal value",cities[2])
cities[2] = 'london'
print("updated value",cities[2])
print(cities)
```
