**Accessing list elements**

```python
cities = ['singapore','newyork','islamabad']
numbers = [45,56,50,64,35]
print(cities[0])
print(numbers[1:4])
```

we can access the elements of lists using indexes/positions.
from above example a list named as cities  and other is numbers.
if we print the list cities but we want only specific elements from the list we will use index like if we write print(cities[0]) the element on index 0 will be printed like singapore was printed above . and the nuumbers are printed from index 1 to 4 excluding 4 because the ending point is always excluded while printing the elements of a list.

**Updating lists**

```python
cities = ['singapore','newyork','islamabad']
print(cities)
print("orignal value",cities[2])
cities[2] = 'london'
print("updated value",cities[2])
print(cities)
```

we can update the values in a list .
 in above example i printed the list. and i wanna update th evalue i=on index 2 so, i printed the value on index 2 (orignal value) then i updated the list and assigned the value 'london' to the list citites[2] and printed it . now the value is updated . also list is updated.

**Deleting a value from a List**

```python
cities = ['singapore','newyork','islamabad']
print("Before deleting", cities)
del cities[2]                          #using del (built-in function)
print('after deleting ', cities)
```

WE can delete elements or values from a list . In above example i declared a list cities[] and then i used a del statment/ function which is a built in function. i deleted the element on index 2 which is the last one. so when i print then list now i will be printed without last element.
