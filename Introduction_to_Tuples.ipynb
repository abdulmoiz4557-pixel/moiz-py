A **tuple** in Python is a built-in data collection type used to store multiple items in a single variable. It is ordered, immutable (unchangeable), and allows duplicate values.tuples are written with round parentheses ().

```python
t1 = (1,2,3,4)    #let a tuple  t1
print(type(t1))
print(t1)
print(t1[0:3])
```

when print the type of t1 it gives me the class 'tuple'.
we can access the elements of tuple by different methods when i print t1 it gives me all elements of the tuple 't1' and when i print the tuple by using index 0:3 it would print the first three elements and excludes the last one.

**Updating and Deleting Tuples**

We cannot delete or update tuples because tuple is immuatble. But we can do it with whole tuple. We can understand it in below given example.

```python
cities = ('lahore','islamabad','multan','swat')
number = (1,2,3,4)
tuple1 = (cities + number)     # adding both tuples
print(tuple1)
```

```python
cities = ('lahore','islamabad','multan','swat')
del cities             #deleting  using del function
print(cities)
```

Deleting tuple is giving error that 'cities is not defined',but cities is in the program ,means that we already deleted that.

**Basic Tuple OPerations**

1- Iteration
This is the first thing that we can do with tuple. we can iterate a tuple. as in following example.

```python
cities = ('lahore','islamabad','multan','swat')
for x in cities:
 print(x)
```

2- Repetition

```python
cities = ('lahore','islamabad','multan','swat')
another = cities * 2
print(another)
```

As in above given example we repeated the tuple twice,in a simple way as we have learnt in strings and lists.
