**Variable types in python**

```python
n=50
print(type(n))
n=50+0.3
print(type(n))
n="seven"
print(type(n))
```

Variable types can be changed in python. in the first line i wwrote n=50. its a literal value. or an integer but in 2nd line i assigned the vale 50+0.3 to n. Its now float. I can also assign other types to a variable like if i write a string as i assigned a string "seven" to n. its now a string. we can also chk the type of a variable by using type() function as a can write print(type(n))
